## User guide
The software consists of three modules listed below, which allow processing of FCS data. Follow the link to learn more about the modules.
- [__*Import PTU*__](Import_PTU.ipynb#Import-PTU)    
- [__*Time-binned correlation*__](Time-binned.ipynb#Time-binned-correlation)    
- [__*FCS fitting*__](FCS_fitting.ipynb#FCS-fitting)    

By default, FcsIT starts in the [__*FCS fitting*__](FCS_fitting.ipynb#FCS-fitting) mode. The other modules can be selected from the Methods menu; see image below.

In the _About_ menu:
  - _Help_ opens this manual in a web browser.
  - _License_ opens the software licence.
  - _Version_ shows the installed software version and indicates when an update is available. Version checking and downloading updates require an internet connection.
