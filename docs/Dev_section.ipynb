## For developers
The FcsIT software consists of three methods. It is, however, possible to extend the existing set of modules. The template for creation own modules is provided in the [templates](https://github.com/TKmist/FcsIT/blob/main/templates/) folder. The installation of the template is as simple as copying the whole `add_on_templeate` folder into the `src/Methods` folder or (once the FcsIT is already installed) into the `FcsIT/Methods` folder.
